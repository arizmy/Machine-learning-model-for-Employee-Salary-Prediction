# Loading dependecies

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import MinMaxScaler 
from sklearn.metrics import root_mean_squared_error, mean_absolute_error #for metric/model evaluation in regression
from sklearn.ensemble import RandomForestRegressor
import pickle
```

# Read  dataset

```python
df = pd.read_excel("employee_data.xlsx")
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
le = LabelEncoder()
```

```python
df.Gender = le.fit_transform(df.Gender)
```

```python
df.Qualifications = le.fit_transform(df.Qualifications)
```

```python
df.Grade = le.fit_transform(df.Grade)
```

```python
df.Department = le.fit_transform(df.Department)
```

```python
df.Employment_Type = le.fit_transform(df.Employment_Type)
```

```python
df.Marital_Status = le.fit_transform(df.Marital_Status)
```

```python
df.Location = le.fit_transform(df.Location)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.columns
```

```python
mm = MinMaxScaler()
```

```python
df.describe()
```

```python
df1 = mm.fit_transform(df)
```

```python
df1
```

```python
df = pd.DataFrame(df1, columns=['Gender', 'Age', 'W_Experience', 'Qualifications', 'Grade', 'Salary',
      'Department', 'Employment_Type', 'Marital_Status', 'Location']) # this convert the df1 into data frame
```

```python
df.drop("Gender", axis=1, inplace=True)
```

```python
df.drop("Location", axis=1, inplace=True)
```

```python
df.drop("Department", axis=1, inplace=True)
```

```python
df
```

```python
X = df.drop("Salary", axis=1)
y = df.Salary 
X_train, X_test, y_train, y_test = train_test_split(X,y, train_size=0.80, test_size=0.20, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
predict = lr.predict (X_test)
```

```python
predict
```

```python
mean_absolute_error(predict,y_test)
```

```python
root_mean_squared_error(predict, y_test)
```

```python
rf = RandomForestRegressor (n_estimators=100 ) #n_estimator constant
```

```python
rf.fit(X_train, y_train)
```

```python
result = rf.predict(X_test)
```

```python
result
```

```python
mean_absolute_error(result, y_test)
```

```python
root_mean_squared_error(result, y_test)
```
